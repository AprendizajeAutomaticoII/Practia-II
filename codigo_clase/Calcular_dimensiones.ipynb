```python
#nn.Conv2d
#Introduzca ancho de entrada: 28
#Introduzca alto de entrada: 28
#Introduzca profundidad de entrada: 1
#Introduzca profundidad de salida: 16
#Introduzca ancho de kernel: 5
#Introduzca alto de kernel: 5
#Introduzca stride: 1
#Introduzca padding de entrada: 2
#N, 1, 28, 28 -> N, 16, 28, 28

import math
ancho_entrada = int(input("Introduzca ancho de entrada: "))
alto_entrada = int(input("Introduzca alto de entrada: "))

profundidad_entrada = int(input("Introduzca profundidad de entrada: "))
profundidad_salida = int(input("Introduzca profundidad de salida: "))

ancho_kernel = int(input("Introduzca ancho de kernel: "))
alto_kernel = int(input("Introduzca alto de kernel: "))

stride = int(input("Introduzca stride: "))
padding_entrada = int(input("Introduzca padding de entrada: "))
ancho_salida = math.floor((ancho_entrada-ancho_kernel+2*padding_entrada)/stride) + 1
alto_salida = math.floor((alto_entrada-alto_kernel+2*padding_entrada)/stride) + 1

print(f"N, {profundidad_entrada}, {ancho_entrada}, {alto_entrada} -> N, {profundidad_salida}, {ancho_salida}, {alto_salida}")

#Ejemplo:
```

```python
#nn.ConvTranspose2d
#Introduzca ancho de entrada: 1
#Introduzca alto de entrada: 1
#Introduzca profundidad de entrada: 64
#Introduzca profundidad de salida: 32
#Introduzca ancho de kernel: 7
#Introduzca alto de kernel: 7
#Introduzca stride: 1
#Introduzca padding de entrada: 0
#Introduzca padding de salida: 0
#N, 64, 1, 1 -> N, 32, 7, 7Introduzca ancho de entrada: 1

ancho_entrada = int(input("Introduzca ancho de entrada: "))
alto_entrada = int(input("Introduzca alto de entrada: "))

profundidad_entrada = int(input("Introduzca profundidad de entrada: "))
profundidad_salida = int(input("Introduzca profundidad de salida: "))

ancho_kernel = int(input("Introduzca ancho de kernel: "))
alto_kernel = int(input("Introduzca alto de kernel: "))

stride = int(input("Introduzca stride: "))

padding_entrada = int(input("Introduzca padding de entrada: "))
padding_salida = int(input("Introduzca padding de salida: "))

alto_salida = (alto_entrada-1)*stride-2*padding_entrada+alto_kernel+padding_salida
ancho_salida = (ancho_entrada-1)*stride-2*padding_entrada+ancho_kernel+padding_salida
print(f"N, {profundidad_entrada}, {ancho_entrada}, {alto_entrada} -> N, {profundidad_salida}, {ancho_salida}, {alto_salida}")
```
